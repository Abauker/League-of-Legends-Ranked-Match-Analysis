# Basic Visualisations

One comparison I wanted to view is one between the win rates of the two teams. To my limited knowledge, it should be roughly 50% for both teams, as supposedly, there is no advantage to being on either team, as the map in which the game takes place is mirrored along the center, giving neither side and advantage.

```python
sns.countplot(data=df, x="winner")
plt.bar_label(plt.gca().containers[0],
              label_type='center')
plt.title("Number of Matches Won by Each Team", fontsize=15)
plt.xticks([0,1], ['Team 1', 'Team 2'])
plt.ylabel("")
plt.show()
```

Interestingly, team 1 has over 600 wins more than team 2. I feel as though this is a significant difference, thought it is less than a 3% difference. There isn't much to say about this, as I am unsure what could cause this.

---

The next piece I want to highlight is the frequency of each character, but with 148 in the game at the time, it is difficult to display them all. To this end, I decided to display the top 39 characters, as well as provide an "other" item showing how often any character not in the top 39 was picked.

Since the Character selection occurs across 10 columns per row, I must first stack all character columns into one variable, champ_counts.

```python
#Grouping all character columns into a single variable.
champ_counts = df[["t1_champ1id","t1_champ2id","t1_champ3id","t1_champ4id","t1_champ5id","t2_champ1id","t2_champ2id","t2_champ3id","t2_champ4id","t2_champ5id"]].stack()
#Printing value counts in order to confirm the above method worked, and to confirm the order of the bar chart.
print(champ_counts.value_counts().head(20))
```

```python
champ_counts = champ_counts.value_counts()

# Keeping the top 39 as a variable to concatenate, while also creating a variable liable for the remaining champions.
top_39 = champ_counts.head(39)
other = champ_counts.iloc[39:].sum()

champ_40 = pd.concat([top_39, pd.Series({"other": other})])
print(champ_40)
```

After making the data easier to present, I decided to place the data into a bar chart, ordered by pickrate, with "other" being pushed to the back. Due to how the above steps were completed, there was no need to  reorder the data to achieve this, so I simply needed to create the plot.

```python
# Plot the above ChampId counts into a bar chart
champ_40.plot(kind='bar', figsize=(10, 6), color='purple')
plt.ylabel('Frequency')
plt.xlabel("ChampId")
plt.gca().set_facecolor('xkcd:mint green')
plt.title('Highest to Lowest Frequency of Champions',
          color='black',
          fontsize=16)
plt.show()
```

After reviewing the above chart, I decided that it would be helpful to have another visualisation of the same data. To this end, I created a pie chart to view the same data. Due to issues with the overall presentation of the chart, I decided to lower the total unique values down from 40, to 20 (including "other"). 

The result may be seen below.

First, the variables used for the above visualisation are slightly tweaked to suit the chart.

```python
# Keeping the top 19 as a variable to concatenate, while also creating a variable liable for the remaining champions.
top_19 = champ_counts.head(19)
other = champ_counts.iloc[19:].sum()

champ_20 = pd.concat([top_19, pd.Series({"other": other})])
```

The chart is created, using the newly tweaked variables.

```python
plt.figure(figsize=(8,8))
plt.pie(champ_20, labels=champ_20.index, autopct='%1.1f%%', colors=sns.color_palette("Set2"), startangle=90)
plt.title("a")
plt.show()
```

From this chart, more meaningful observations can be made, such as the fact that the top 19 champions dominate almost a third of the total pick rate.

While I am not displaying the 40 unique values currently, when I did so, the "other" section held only 47% of all character selections, which may be interesting to note.

I decided to leave the basic visualisations at this, and handle other visualisations later on, as they will typically be more advanced than this.
