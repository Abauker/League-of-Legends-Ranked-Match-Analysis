# Understanding the Data

Briefly looking over the data as a whole

```python
# Displaying the shape of the dataset, as rows and columns.
print(df.shape)

# Displaying the information of the dataset, including data types and non-null counts.
print(df.info())
```

I find it necessary to review what exactly the columns display, since the naming can be rather vague if one is unfamiliar with the game.

Firstly I will state my primary goal with this data - To predict the winner of a match using given data points.
I am focused primarily on using the T1 and T2 (team 1 and team 2) champion ID data to do so, as the champions (characters each player uses) are generally what determines the outcome of a match, outside of elements such as personal skill of players.

Many of the columns are, at least for my purposes, rather obtuse. Datapoints in columns 0 and 1 refer to what I am going to class as metadata - internal information related to the game which shouldn't have any impact on the data. Datapoints given in the "ban" columns can fall under the same category, as they shouldn't have any impact on the match itself - The bans simply reflect characters that players do not wish to play against, and are declared prior to a match beginning. An additional reduntant datapoint is the SeasonID, which only contains "9", making it irrelevant, as there is never a change in it.

It can be argued that game duration is a useful data point, as some characters perform better in lengthier matches, however I primarily want to focus on predictions at the start of a match, rather than having knowledge of how long a match lasts - something decided by player action during a match. Following this, Columns such as 5 - 10 are similiarly useless to my specific purpose, as they provide information on which team was able to complete the respective objectives during the game.

Ordinarily, the characters each would have their own names, but within this dataset, they were converted to their internal ID numbers - something that I will not reverse, as I find it unecessary to know the names of each character present in the data.

# Addressing Missing / Duplicated Values

First, I will confirm that there are missing values that will need to be addressed in the ETL

```python
# Allowing all rows to be displayed at once.
pd.set_option('display.max_rows', None)
# Check for missing values
df.isnull().sum()
```

From this, it is clear that there are no missing values to be handled later.

I will now do the same for Duplicated Values - To do so, I will search for duplicated values for the "gameId" column, as it highlights each instance of each unique match within the dataset.

```python
# Check for duplicates
duplicates = df.duplicated('gameId').sum()
print(f'Duplicated values - {duplicates}')
```

From this, it is clear that there are duplicated values to be handled within the ETL.
